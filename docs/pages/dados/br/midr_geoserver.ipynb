# Ministério da Integração e do Desenvolvimento Regional (MDR)


```python
import io

import geopandas as gpd
from owslib.wfs import WebFeatureService
```

```python
wfs = WebFeatureService(
    #
    url='https://geoserver.mdr.gov.br/geoserver/ows',
    version='2.0.0',
)
```

<br>


----

## Listas *Layers*

```python
list(wfs.contents)
```

<br>

-----

## Obter *Layers*

```python
# Obter os dados no formato GeoJSON (ou outro formato suportado)
response = wfs.getfeature(
    typename='SGDGeo:georegiao',
    # bbox=(173700, 440400, 178700, 441400),
    # srsname='EPSG:28992'
    # srsname='EPSG:4326',
    srsname='EPSG:4674',
    outputFormat='application/json',
)
response
```

<br>

----

## Geopandas

```python
gdf = gpd.read_file(filename=io.BytesIO(response.read()))
gdf.info()
gdf.head(2)
```
