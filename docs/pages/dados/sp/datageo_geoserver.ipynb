# DataGeo

Especificamente para o DataGeo observei que o WFS está desabilitado. Contudo, pela interface, é possível fazer o download das informações (e isso só é possível com WFS hablitado).

- https://datageo.ambiente.sp.gov.br/geoserver/web
- https://datageo.ambiente.sp.gov.br/geoserver/ows
- https://datageo.ambiente.sp.gov.br/geoserver/datageo/ows
- https://201.55.44.221/


```python
import geopandas as gpd
from owslib.wfs import WebFeatureService
```

```python
wfs = WebFeatureService(
    # Serviços
    # url='https://datageo.ambiente.sp.gov.br/geoserver/ows', # WMS apenas
    url='https://datageo.ambiente.sp.gov.br/geoserver/datageo/ows',  # WFS habilitado
    version='2.0.0',
)
```

<br>


----

## Listas *Layers*

```python
list_layers = list(wfs.contents)
list_layers.sort()
list_layers
```

<br>

-----

## Obter *Layers*

```python
# Obter os dados no formato GeoJSON (ou outro formato suportado)
response = wfs.getfeature(
    typename='datageo:VMW_UC_ESTADUAL_FE_Guarulhos_POL',
    # bbox=(173700, 440400, 178700, 441400),
    # srsname='EPSG:28992'
    # srsname='EPSG:4326',
    srsname='EPSG:4674',
    outputFormat='application/json',
)
response
```

<br>

----

## Geopandas

```python
gdf = gpd.read_file(filename=response)
gdf.crs
```

```python
gdf.info()
```
