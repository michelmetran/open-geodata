# Defesa Civil

[![Repo](https://img.shields.io/badge/GitHub-repo-blue?logo=github&logoColor=f5f5f5)](https://github.com/open-geodata/sp_defesacivil_divadmin)

O repositório [open-geodata/sp_defesacivil_divadmin](https://github.com/open-geodata/sp_defesacivil_divadmin) apresenta uma forma de obter detalhes sobre as _Coordenadorias Regionais de Proteção e Defesa Civil_ (REPDEC) da Defesa Civil do Estado de São Paulo.
Usando o pacote [open-geodata](https://github.com/michelmetran/open-geodata) é possível ler o conteúdo do repositório, com auxílio do [pooch](https://www.fatiando.org/pooch/latest/).


```python
import tempfile
from pathlib import Path

import open_geodata as geo
```

<br>

## _Layers_

Abaixo listamos todos os _layers_/dados disponíveis no repositório.


```python
db = geo.data.DB(db='sp_defesa_civil')
db.list_data
```

<br>

E selecionamos o que desejamos visualizar.

```python
gdf = geo.data.load_dataset(db='sp_defesa_civil', name='geo.sp_defesa_civil')

# Results
gdf.info()
gdf.head(2)
```

```python
gdf.explore(column='repdec_regiao')
```

<br>
 
Criamos uma pasta temporária apenas para demonstrar como podemos salvar o resultado.


```python
temp_path = Path(tempfile.gettempdir()) / 'open_geodata'
temp_path.mkdir(exist_ok=True)
```

<br>

E salvamos em formatos distintos.

```python
# Salva em Geopackage
gdf.to_file(
    temp_path / 'sp_cetesb.gpkg',
    layer='sp_cetesb',
    driver='GPKG',
)


# Salva em Geojson
gdf.to_file(
    temp_path / 'sp_cetesb.geojson',
    driver='GeoJSON',
    encoding='utf-8',
)
```
